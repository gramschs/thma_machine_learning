# 4.3 Scatterplots und Scattermatrix

Bei der Datenvisualisierung geht es darum, Daten durch eine Grafik so
aufzubereiten, dass Muster oder Unregelmäßigkeiten in den Daten entdeckt werden
können. Dabei kann die visuelle Darstellung der Daten helfen, Muster in den
Daten zu entdecken, aber sie kann auch irreführend sein. Abhängig davon, wie die
Art der Daten beschaffen ist, die wir visualisieren wollen, gibt es verschiedene
Darstellungsformen, die sogenannten **Diagrammtypen**. Im Folgenden betrachten wir
die Diagrammtypen

* Scatterplot und
* Scattermatrix.

Danach beschäftigen wir uns mit der Gestaltung bzw. dem Styling von Diagrammen.

## Styling von Scatterplots

Die Voreinstellungen von Plotly sind bereits sehr gut gewählt, so dass ohne
weitere Optionen bereits gut aussehende und informative Diagramme erstellt
werden können. Eine Möglichkeit, durch das Styling der Diagramme
Zusatzinformationen zu visualisieren, bietet die Option `text=`. Wir verwenden
den Zeilenindex als Text, der in dem Attribut `.index` gespeichert ist.

```python
diagramm = px.scatter(daten, x = 'Kilometerstand (km)', y = 'Preis (Euro)',
                      # TODO: ???  (Zeilenindex als Text an die Kreise schreiben)
                      )
diagramm.show()
```

An jedem kleinen Kreis wird nun zusätzlich die Auto-ID eingeblendet. Leider
überschreibt der Text den Kreis selbst. Das kann nachträglich geändert werden,
indem die Textposition relativ zu den Kreisen auf einen anderen Wert gesetzt
wird. Die einzelnen Bestandteile eines Plotly-Express-Diagramms heißen
**trace**. Sie werden durch `update_traces()` aktualisiert. Wir möchten, dass
die Position der Texte oberhalb der Kreise ist, aber dennoch zentriert dazu.
Durch das Argument `textposition='top center'` erreichen wir dieses Ziel, wie
der folgende Scatterplot zeigt.

```python
diagramm = px.scatter(daten, x = 'Kilometerstand (km)', y = 'Preis (Euro)',
                      text=daten.index)
diagramm.update_traces(textposition='top center')
diagramm.show()
```

Als Nächstes möchten wir weitere Zusatzinformationen in das Diagramm packen.
Nicht immer ist es sinnvoll, so viele Zusatzinformationen in ein Diagramm zu
bringen, da damit das Publikum auch schnell überfordert werden kann. Daher
sollte gut überlegt werden, ob die beiden nächsten Möglichkeiten gleichzeitig
genutzt werden sollen.  

Die **Farbe** ist eine weitere Möglichkeit, Zusatzinformationen zu
visualisieren. Wie alt ist ein Auto? Hat es ebenfalls einen Einfluss auf den
Verkaufspreis? Wir nutzen das Jahr der Erstzulassung, um das Alter der Autos
abzuschätzen. Die Anweisung an Python, die Kreise des Scatterplots nach der
Erstzulassung einzufärben, wird durch das optionale Argument `color='Jahr'`
gegeben.

```python
diagramm = px.scatter(daten, x = 'Kilometerstand (km)', y = 'Preis (Euro)',
                      text=daten.index,
                      # TODO: ???  (Kreise nach dem Jahr einfärben)
                      )
diagramm.update_traces(textposition='top center')
diagramm.show()
```

Da 'Jahr' ein numerisches Merkmal ist, verwendet Plotly eine kontinuierliche
Farbskala. Welche Farbe für ältere oder jüngere Fahrzeuge steht, ist an der
eingeblendeten Farbskala abzulesen. Wir finden die helleren Farben eher links im
Diagramm. Also könnten wir die Vermutung aufstellen, dass das Jahr der
Erstzulassung und damit das Alter der Fahrzeuge auch etwas mit dem
Kilometerstand zu tun zu haben. Tendenziell scheint für diese 10 Autos zu
gelten: Je jünger das Fahrzeug ist, desto weniger Kilometer wurde es bisher
gefahren.

Als zweite Möglichkeit, Zusatzinformationen direkt mit den Datenpunkten im
Scatterplot zu visualisieren, dient die **Größe der Kreise**. Mit dem optionalen
Argument `size=` wird sie gesteuert. Wiederum verwenden wir einen Spaltenindex
als Argument. Die Leistung könnte erfahrungsgemäß ebenfalls den Verkaufspreis
beeinflussen. Also setzen wir `size='Leistung (PS)'` und betrachten das so
erweiterte Diagramm.

```python
diagramm = px.scatter(daten, x = 'Kilometerstand (km)', y = 'Preis (Euro)',
                      text=daten.index, color='Jahr', size='Leistung (PS)')
diagramm.update_traces(textposition='top center')
diagramm.show()
```

Das Auto 'BMW Nr. 1', das uns schon zuvor aufgefallen ist, weil der Preis trotz
des mittleren Kilometerstandes recht hoch ist, scheint besonders viel PS zu
haben. Vielleicht erklärt das den hohen Preis?

Als letzte Styling-Möglichkeit betrachten wir den **Titel**. Jedes Diagramm muss
eindeutig beschriftet sein. Das kann durch einen aussagekräftigen Titel oder
eine Abbildungsunterschrift erfolgen. Der Titel wird mit dem Argument `title=`
gesetzt.

```python
diagramm = px.scatter(daten, x = 'Kilometerstand (km)', y = 'Preis (Euro)',
                      text=daten.index, color='Jahr', size='Leistung (PS)',
                      title='Verkaufsdaten von 10 Autos (Quelle: AutoScout24.de)')
diagramm.update_traces(textposition='top center')
diagramm.show()
```

## Scattermatrix

Unsere Tabelle hat sieben Spalten mit numerischen Werten: Jahr, Preis (Euro),
Leistung (kW), Leistung (PS), Verbrauch (l/100 km), Verbrauch (g/km) und
Kilometerstand (km). Wenn wir jeweils zwei unterschiedliche Merkmale miteinander
vergleichen, kann

* das erste Merkmal mit sechs anderen Merkmalen verglichen werden,
* das zweite Merkmal mit fünf anderen Merkmalen verglichen werden (mit dem
  ersten Merkmal wurde es bereits verglichen),
* das dritte Merkmal mit vier anderen Merkmalen verglichen werden (mit dem
  ersten und dem zweiten wurde es bereits verglichen), usw.

Damit ergeben sich

$$6 + 5 + 4 + 3 + 2 + 1 = 21$$

Paare von Merkmalen. Die Scattermatrix vereinfacht das Zusammenstellen dieser
Kombinationen. Dazu legen wir erst eine Liste mit den Spaltenlabels an, die in
die Scattermatrix aufgenommen werden sollen. Danach erzeugen wir mit der
Funktion `px.scatter_matrix()` die gewünschten Kombinationen. Als erstes
Argument werden die Daten aus der Tabelle übergeben, dann folgt die Liste der
ausgewählten Spalten als Argument für den Parameter `dimensions=`.

Es werden 49 Diagramme angezeigt, die allerdings kaum lesbar sind. Warum 49 und
nicht 21? Jedes Paar von Merkmalen wird zweimal visualisiert: einmal mit Merkmal
1 auf der x-Achse und Merkmal 2 auf der y-Achse sowie einmal umgekehrt. Darüber
hinaus enthält die Diagonale sieben Felder, in denen ein Merkmal gegen sich
selbst geplottet wird. Da das Diagramm so kaum lesbar ist, reduzieren wir die
Auswahl weiter und nehmen nur die ersten vier Eigenschaften, bevor wir die
Interpretation der Scattermatrix angehen.

```python
auswahl = ['Jahr', 'Preis (Euro)', 'Leistung (kW)', 'Leistung (PS)']

diagramm = px.scatter_matrix(daten, dimensions=auswahl)
diagramm.show()
```

Auf der Diagonalen befinden sich Scatterplots, bei denen dasselbe Merkmal auf
der x- und auf der y-Achse aufgetragen ist. Daher liegen die Kreise dort
zwangsläufig auf der Winkelhalbierenden. Diese Diagramme liefern keinen
zusätzlichen Hinweis auf Zusammenhänge zwischen unterschiedlichen Merkmalen.
Möchten wir die Verteilung eines einzelnen Merkmals untersuchen, verwenden wir
besser Histogramme oder Boxplots. Daher beachten wir die Diagonalen nicht
weiter.

Betrachten wir den Scatterplot Leistung (kW) vs. Leistung (PS), so stellen wir
fest, dass die Kreise auf einer Geraden liegen. Tatsächlich ist das nicht
verwunderlich, da die beiden Spalten ja dieselbe physikalische Größe
beschreiben, die nur in zwei verschiedenen Einheiten angegeben wird. 1 kW sind
ungefähr 1.36 PS. Die Scattermatrix zeigt uns nun (wenn wir es nicht schon
vorher wussten), dass wir nur eine der beiden Spalten brauchen. Diese Spalte
könnten wir für die weitere Datenexploration ignorieren.

```python
auswahl = ['Jahr', 'Preis (Euro)', 'Leistung (kW)']

diagramm = px.scatter_matrix(daten, dimensions=auswahl)
diagramm.show()
```

Als letztes Interpretationsbeispiel betrachten wir die zweite Zeile der Matrix,
wo der Preis (Euro) auf der y-Achse aufgetragen ist. Betrachten wir den ersten
Scatterplot der zweiten Zeile, so scheinen das Jahr und der Verkaufspreis keinen
besonderen Zusammenhang zu haben. Beim dritten Scatterplot in der zweiten Zeile,
scheint es aber ein Muster zu geben. Mit wachsender Leistung scheint auch der
Verkaufspreis zu steigen. Die Scattermatrix hilft also gerade zu Beginn der
Datenexploration schnell, interessante Zusammenhänge zwischen einzelnen
Merkmalen aufzudecken, die dann durch einzelne Scatterplots näher untersucht
werden können.
